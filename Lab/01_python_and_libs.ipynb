```python
n = 10000
data2 = np.random.normal(0, 1, size=(n, 2))

img1 = np.random.rand(64, 64, 1)
img3 = np.random.rand(64, 64, 3)
```

```python
fig, axs = plt.subplots(2, 2, figsize=(6, 6))

ax = axs[0, 0]
ax.plot(data2)
ax.set_xlabel('index')
ax.set_ylabel('x')
ax.set_title('index vs x')
ax.grid()
ax.legend()

ax = axs[0, 1]
ax.scatter(data2[:, 0], data2[:, 1], alpha=0.05)
ax.set_xlabel('coord0')
ax.set_ylabel('coord1')
ax.set_title('scatterplot')
ax.grid()
ax.legend()

ax = axs[1, 0]
ax.imshow(img1)

ax = axs[1, 1]
ax.imshow(img3)

plt.savefig("image.pdf")
plt.savefig("image.jpg")
```

```python
xd = range(100)
yd = range(100)

xy = np.array(np.meshgrid(xd, yd)).reshape(2, -1)

xx, yy = np.meshgrid(xd, yd)
zz = np.sin(xx / 4) + np.cos(yy / 4) + 2 * np.exp(1 / (xx + yy + 1) / 2)
```

```python
plt.imshow(zz)
```

```python
plt.pcolormesh(xx, yy, zz)
```

```python
plt.contour(xx, yy, zz)
```

```python
plt.contourf(xx, yy, zz)
```

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter_3d(df, x='sepal_length', y='sepal_width', z='petal_width',
              color='species')
fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure(
    data=[
        go.Surface(
            x=xx,
            y=yy,
            z=zz
        )
    ]
)

fig.update_layout(
    title="3D Surface",
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z",
    ),
    width=900,
    height=700,
)

fig.show()
```
